### Question 1 (Easy) — `can_vote`

**Expected:** `can_vote(age)` should return `True` for anyone aged **18 or older**, and `False` otherwise.

- `can_vote(25)` ➞ `True`
- `can_vote(18)` ➞ `True`
- `can_vote(10)` ➞ `False`

It runs with no error — check every output against the expected values.

```python
def can_vote(age):
    if age > 18:
        return True
    else:
        return False

print(can_vote(25))
print(can_vote(18))
print(can_vote(10))
```

### Question 2 (Medium) — `shipping_fee`

**Expected:** up to and including 1 kg costs `5000`; over 1 kg up to and including 5 kg costs `12000`; over 5 kg up to and including 20 kg costs `25000`; over 20 kg costs `40000`.

- `shipping_fee(3)` ➞ `12000`
- `shipping_fee(20)` ➞ `25000`
- `shipping_fee(30)` ➞ `40000`

Check **all three** calls. One input falls through every branch.

```python
def shipping_fee(weight_kg):
    if weight_kg <= 1:
        return 5000
    elif weight_kg <= 5:
        return 12000
    elif weight_kg < 20:
        return 25000
    elif weight_kg > 20:
        return 40000

print(shipping_fee(3))
print(shipping_fee(20))
print(shipping_fee(30))
```

### Question 3 (Hard) — `sum_by_twos`

**Expected:** `sum_by_twos(n)` returns `n + (n-2) + (n-4) + ...` down to `1` or `2` (whichever it reaches).

- `sum_by_twos(6)` ➞ `12` (6 + 4 + 2)
- `sum_by_twos(7)` ➞ `16` (7 + 5 + 3 + 1)

The first call works. What happens on the second, and why does the first one get away with it?

```python
def sum_by_twos(n):
    if n == 0:
        return 0
    return n + sum_by_twos(n - 2)

print(sum_by_twos(6))
print(sum_by_twos(7))
```
