### Question 1 (Easy) — `has_word`

**Expected:** `has_word(sentence, word)` should return `True` only if `word` appears in `sentence` as a **whole word**.

- `has_word('I love my cat', 'cat')` ➞ `True`
- `has_word('this is a scattered list', 'cat')` ➞ `False`
- `has_word('the category is full', 'cat')` ➞ `False`

The first call is right. Why does Python think there is a cat in the other two sentences? *(You can assume the sentences have no punctuation.)*

````python
def has_word(sentence, word):
    return word in sentence

print(has_word('I love my cat', 'cat'))
print(has_word('this is a scattered list', 'cat'))
print(has_word('the category is full', 'cat'))
````

### Question 2 (Medium) — `long_words`

**Expected:** `long_words(sentence, n)` should return a **list of the words** that have **more than** `n` letters, in the order they appear.

- `long_words('Python is a fun language', 4)` ➞ `['Python', 'language']`
- `long_words('I am ok', 5)` ➞ `[]`

The second call looks fine. Run the first one and look closely at what ends up in the list.

````python
def long_words(sentence, n):
    result = []
    for word in sentence.split():
        if len(word) > n:
            result += word
    return result

print(long_words('Python is a fun language', 4))
print(long_words('I am ok', 5))
````

### Question 3 (Hard) — `count_lines_with`

**Expected:** `count_lines_with(lines, word)` gets a list of log lines and should return how many lines **contain** `word` anywhere.

- `count_lines_with(['error: disk full', 'all good', 'login error', 'ok'], 'error')` ➞ `2`
- `count_lines_with(['ok', 'fine'], 'error')` ➞ `0`
- `count_lines_with(['error 1', 'error 2'], 'error')` ➞ `2`

**Check all three calls.** The first answer is only off by one, so it's easy to miss — but the other two are wrong in *opposite* directions. What does `.find()` return when the word is at the very start of the line? And when it isn't there at all?

````python
def count_lines_with(lines, word):
    count = 0
    for line in lines:
        if line.find(word):
            count += 1
    return count

print(count_lines_with(['error: disk full', 'all good', 'login error', 'ok'], 'error'))
print(count_lines_with(['ok', 'fine'], 'error'))
print(count_lines_with(['error 1', 'error 2'], 'error'))
````
