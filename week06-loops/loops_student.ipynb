### Concept Recap (quick — no slides needed today)

| | `for` loop | `while` loop |
|---|---|---|
| Use when | you know how many times (or what to walk through) | you repeat *until a condition changes* |
| Looks like | `for i in range(5):` | `while n > 0:` |
| Danger | off-by-one: `range(5)` is `0..4` | **infinite loop** if nothing changes the condition |

```python
total = 0                 # 1. set up BEFORE the loop
for n in range(1, 5):
    total = total + n     # 2. update INSIDE the loop
return total              # 3. use the result AFTER the loop
```

- `break` leaves the loop completely; `continue` skips to the next pass; `return` leaves the whole **function**.
- `range(a, b)` goes from `a` up to **but not including** `b`; `range(a, b, step)` counts by `step` (negative to count down).

## 3. Debugging Round — Broken Loops

Each function is *already written* and has exactly one bug. Run it, compare with the expected value, and fix the bug in that cell.

### Debug 1 — `sum_to_n`

**Expected:** `sum_to_n(4)` should return `10` (1+2+3+4) and `sum_to_n(1)` should return `1`.

It runs without an error, but the answer is too small. Look at the numbers `range` actually produces.

````python
def sum_to_n(n):
    total = 0
    for i in range(1, n):
        total = total + i
    return total

print(sum_to_n(4))
print(sum_to_n(1))
````

### Debug 2 — `countdown`

**Expected:** `countdown(3)` should print `3`, `2`, `1`, then `Liftoff!`.

> **Careful:** read the code *before* running it and predict what will happen. If the cell never finishes, press the ■ **Stop** button (Colab) or **Kernel → Interrupt** (Jupyter).

````python
def countdown(n):
    while n > 0:
        print(n)
    print("Liftoff!")

countdown(3)
````

### Debug 3 — `factorial`

**Expected:** `factorial(5)` should return `120` and `factorial(3)` should return `6`.

It runs without an error, but gives `5` and `3`. Where does `result` get its value on each pass?

````python
def factorial(n):
    for i in range(1, n + 1):
        result = 1
        result = result * i
    return result

print(factorial(5))
print(factorial(3))
````

### Debug 4 — `count_vowels`

**Expected:** `count_vowels("education")` should return `5` and `count_vowels("sky")` should return `0`.

It returns `1` for the first one. Check the indentation of the last line of the loop.

````python
def count_vowels(s):
    count = 0
    for ch in s:
        if ch in "aeiou":
            count = count + 1
        return count

print(count_vowels("education"))
print(count_vowels("sky"))
````

## What We Practiced

- Predicting loops line by line: `range(start, stop, step)` stops *before* `stop`, and indentation decides what is inside the loop
