### Concept Recap: What Is a Function?

*(Pause here for Prof. Kabir's slides on this, then use this cell as your notes anchor.)*

- **Function definition** — `def name(parameters):` followed by an indented body. Defining a function does **not** run it.
- **Function call** — `name(arguments)` actually runs the function's body.
- **Parameter vs. argument** — a parameter is the name in the `def` line; an argument is the actual value you pass in when you call it.
- **Built-in functions** — Python already ships with functions like `print()`, `len()`, `type()`, `round()`. `free_fall_distance` above is a **user-defined** function: one you wrote yourself.
- **Default parameters** — a parameter can have a default value (`def f(x, y=10):`), so a caller can leave that argument out. Demo 1b below uses one.
- **`return` vs. `print`** — `print` only displays a value; `return` hands the value back to whoever called the function so it can be stored, reused, or passed along.
- **Local variable** — a variable created inside a function only exists while that function is running.
