## 1. Foundations — Code Blocks, Indentation, `return` vs `print`

**Remember:**

- A **block** starts after a line that ends with `:` (`def`, `if`, `elif`, `else`, `for`, `while`).
- Every line indented under it (4 spaces) is **inside** that block. The first line back at the old indentation is **outside**, so the block has ended.
- Indentation is part of the code, not decoration. Moving a line left or right changes **when** and **how many times** it runs. Messy indentation gives `IndentationError`.
- `print` **shows** a value on the screen. `return` **hands a value back** to whoever called the function.
- A function with no `return` gives back `None`. So `x = f()` stores `None`, and `print(f())` prints whatever `f` printed, then `None`.

### Q1 — Predict

What is printed? Write it down **before** you run the cell.

````python
total = 0
for i in range(1, 4):
    total += i
    print(total)
print("done", total)
````

**Answer:**

```
1
3
6
done 6
```

**Why:** `print(total)` is **inside** the loop, so it runs 3 times. `print("done", total)` is back at the left edge, so it runs **once**, after the loop finishes.

### Q4 — Debug 🐞

**Goal:** `count_vowels` should count **all** the vowels in a word.

a) What does it print **as written**?  b) Find the bug and fix it in the cell below, then re-run.

````python
def count_vowels(word):
    count = 0
    for ch in word:
        if ch in "aeiou":
            count += 1
        return count

print(count_vowels("banana"))   # expected 3
````

**Answer:** It prints:

```

```

## 2. Basics — Types and a Few Arithmetic Operators

**Remember:**

- `/` **always** gives a float (`6 / 3` is `2.0`). `//` is floor division, which rounds **down** toward −∞, so `-7 // 2` is `-4`. `%` is the remainder.
- `**` is done before `*` and `/`, and it is evaluated **right to left**: `2 ** 3 ** 2` is `2 ** 9`.
- `+` on two strings **joins** them (`"3" + "4"` is `"34"`). `input()` always gives a string, so use `int()` or `float()` to turn it into a number.
- `print(a, b, sep="-", end="!")`: `sep` goes **between** the values (default is a space) and `end` goes **after** the last one (default is a new line).

## 3. Functions

**Remember:**

- A function **stops** at the first `return` it reaches, and code after that `return` never runs.
- A **default** value is used only when you leave that argument out.
- Arguments are matched **by position** unless you name them (`f(b=1, a=2)`).
- A variable created inside a function is **local**. Changing it does not change a variable with the same name outside.
- A function's result can be passed straight into another function: `square(change(1))`.

### Q11 — Debug 🐞

## 4. Conditionals

**Remember:**

- `if / elif / else`: Python checks **top to bottom** and runs **only the first** branch that is True. Then it skips the rest.
- Several **separate** `if`s are each checked on their own, so more than one can run.
- `=` **stores** a value; `==` **compares** two values.
- `0`, `0.0` and `""` count as **False** inside an `if`. Any other number and any non-empty string count as **True**.
- `x == "a" or "b"` is **always** True. Write `x == "a" or x == "b"`.

## 5. `for` Loops and `range`

- `range(stop)` starts at **0** and stops **before** `stop`.
- `range(start, stop, step)`: with a negative step it counts down, and `stop` is still **not included**. If the range is empty, the loop body never runs.
- After the loop ends, the loop variable keeps its **last** value.
- Accumulators (`total = 0`, `count = 0`) are set up **before** the loop, not inside it.
- `for ch in "text"` visits each character, one at a time.

## 6. `while`, `break` / `continue`, Nested Loops

**Remember:**

- `while` keeps repeating **as long as** its condition is True, so something inside must eventually make it False.
- `break` exits the **innermost** loop immediately. `continue` skips the rest of the body and jumps to the **next** pass.
- In a `while` loop, if `continue` skips the line that updates the counter, you get an **infinite loop**.
- In nested loops, the inner loop runs **all the way through** for every pass of the outer loop.
- `n % 10` gives the last digit of `n`; `n //= 10` removes it.

### Q23 — Predict

What is printed? Write it down **before** you run the cell.

````python
n = 4721
digits = 0
total = 0
while n > 0:
    total += n % 10
    n //= 10
    digits += 1
print(digits, total, n)
````

**Answer:**

```
4 14 0
```

**Why:** The loop takes off one digit per pass: 1, 2, 7, then 4. That is 4 digits, and their sum is 14. The loop stops when `n` becomes 0.

### Q24 — Predict

**Answer:**

```

```

**Why:** `continue` skips the `print` for 3. `break` stops the whole loop at 6, before it is printed. `i` stays 6 after the loop.

### Q26 — Debug 🐞

**Goal:** Print the odd numbers from 1 to 9. ⚠️ **Do not run the buggy version**, because it never stops. Trace it on paper instead.

a) What happens when this runs?  b) Fix it.

```

while i <= 9:
    if i % 2 == 0:
        continue
    print(i)
    i += 1
```

**Answer:** It prints `1` and then **never stops**.

**Bug:** **Infinite loop:** `continue` skips `i += 1`.

## 7. Bitwise Operators and Number Systems

**Remember:**

- `0b` means binary, `0o` octal and `0x` hex. `bin()`, `oct()` and `hex()` convert a number to those forms, and `int("1011", 2)` converts back.
- `&` gives 1 where **both** bits are 1. `|` gives 1 where **either** is 1. `^` gives 1 where **exactly one** is 1.
- `~x` is `-x - 1`. `x << k` is `x * 2**k`. `x >> k` is `x // 2**k`.
- Permission flags: `p |= F` **gives** a flag, `p & F` **checks** it, `p ^= F` **flips** it and `p &= ~F` **removes** it.
- `and` / `or` work on **True/False**; `&` / `|` work on **bits**. And `^` is **not** power: `2 ^ 3` is `1`. Power is `**`.

### Q28 — Predict

What is printed? Write it down **before** you run the cell.

````python
print(0b1011)
print(bin(13))
print(hex(255), oct(8))
print(int("110", 2) + 0x10)
````

**Answer:**

```
11
0b1101
0xff 0o10
22
```

**Why:** `1011` = 8 + 2 + 1 = 11. 13 = 8 + 4 + 1, which is `1101`. 255 = 15 × 16 + 15, which is `ff`. 8 is `10` in base 8. `110` in binary is 6, `0x10` is 16, and 6 + 16 = 22.

### Q32 — Debug 🐞

**Goal:** `can_write(perms)` should return `True` when the WRITE bit is set.

a) What does it print **as written**?  b) Find the bug and fix it in the cell below, then re-run.

````python
WRITE = 2

def can_write(perms):
    if perms and WRITE:
        return True
    return False

print(can_write(6))   # 6 = 0b110, has WRITE -> expected True
print(can_write(4))   # 4 = 0b100, no WRITE  -> expected False
````

**Answer:** It prints:

```
True
True
```

**Bug:** **`and` used instead of `&`.**

**Why:** `perms and WRITE` only asks *“are both numbers non-zero?”*. 4 and 2 are both non-zero, so it says True. `perms & WRITE` compares the **bits**: `100 & 010 = 000`, which is 0, so it counts as False.

**Fixed version:**

````python
WRITE = 2

def can_write(perms):
    if perms & WRITE:
        return True
    return False

print(can_write(6))   # 6 = 0b110, has WRITE -> expected True
print(can_write(4))   # 4 = 0b100, no WRITE  -> expected False
````

### Q36 — Debug 🐞

**Goal:** `is_prime(n)` should return `True` only if `n` has no divisors other than 1 and itself.

a) What does it print **as written**?  b) Find the bug and fix it in the cell below, then re-run.

````python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
        else:
            return True

print(is_prime(9))   # expected False
print(is_prime(7))   # expected True
````

**Answer:** It prints:

```
True
True
```

**Bug:** **`return True` inside the loop** (in the `else`).

**Why:** The loop only ever tests `d = 2`: 9 % 2 is not 0, so the function says “prime” right away. You can say *prime* only **after** the loop has tried every `d`. Bonus: for `is_prime(2)` the buggy version returns `None`, because the loop never runs.

**Fixed version:**

````python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True               # only after checking EVERY d

print(is_prime(9))   # expected False
print(is_prime(7))   # expected True
````

## What We Practiced

| topic | the trap to watch for |
|---|---|
| Blocks and indentation | One extra or missing indent changes **how many times** a line runs |
| `return` vs `print` | No `return` gives `None`, and `print(f())` can print `None` |
| Arithmetic | `/` gives a float, `//` rounds down, `**` comes first and goes right to left |
| Functions | Code after `return` is dead; arguments are matched by position; variables inside are local |
| Conditionals | Only the first true `elif` runs; `x == "a" or "b"` is always True |
| `for` / `range` | `stop` is not included; reset accumulators **before** the loop; the loop variable keeps its last value |
| `while` / `break` / `continue` | Update the counter before `continue`; `break` exits only the inner loop |
| Bitwise | Line up the bits; `~x = -x-1`; `&` is not `and`, and `^` is not power |
