## Question 1 (Strings, Output)

What is printed?

````python
s = 'Python'
print(s[::-1])
print(s[::2])
````

A) nohtyP, then Pto

B) nohtyP, then yhn

C) Python, then Pto

D) nohty, then Pth

Answer: A

A step of -1 reads the string backwards. A step of 2 takes every other character: indexes 0, 2 and 4 (P, t, o).

Real output:
```
nohtyP
Pto
```

## Question 2 (Strings, Output)

What is printed?

````python
word = 'coding'
result = ''
for ch in word:
    if ch in 'aeiou':
        result = result + ch.upper()
    else:
        result = result + ch
print(result)
````

A) CODING

B) OI

C) coding

D) cOdIng

Answer: D

The loop copies every letter, but the vowels o and i are added in upper case. ch.upper() returns a new string; word itself never changes.

Real output:
```
cOdIng
```

## Question 3 (Strings, Debug)

capitalize_first('inha') should return 'Inha', but it crashes with TypeError. Which change fixes it?

````python
def capitalize_first(word):
    word[0] = word[0].upper()
    return word

print(capitalize_first('inha'))
````

A) Change line 2 to word[0] = word.upper()

B) Change line 3 to return word.upper()

C) Replace lines 2–3 with return word[0].upper() + word[1:]

D) Change line 2 to word[1] = word[0].upper()

Answer: C

Strings are immutable: you can't assign to word[0]. Build a new string from the upper-case first letter and the slice of the rest. (Option B returns 'INHA'.)

Real output:
```
TypeError: 'str' object does not support item assignment
```

After the fix it prints:
```
Inha
```

## Question 4 (Lists, Output)

What is printed?

````python
a = [1, 2, 3]
b = a
c = a.copy()
b.append(4)
c.append(5)
print(a)
print(len(b), len(c))
````

A) [1, 2, 3], then 4 4

B) [1, 2, 3, 4], then 4 4

C) [1, 2, 3, 4, 5], then 5 5

D) [1, 2, 3, 4], then 4 5

Answer: B

b = a gives the same list a second name, so appending to b changes a. c is a real copy, so c.append(5) doesn't touch a. Both b and c have 4 items.

Real output:
```
[1, 2, 3, 4]
4 4
```

## Question 5 (Lists, Debug)

remove_fails([50, 40, 90, 30]) should return [90], but it returns [40, 90]. Which change fixes it?

````python
def remove_fails(scores):
    for s in scores:
        if s < 60:
            scores.remove(s)
    return scores

print(remove_fails([50, 40, 90, 30]))
````

A) Change < to <=

B) Move return scores inside the loop

C) Change scores.remove(s) to scores.pop(s)

D) Change line 2 to for s in scores.copy():

Answer: D

Removing from a list while looping over it shifts the items, so the loop skips the item after each removal (here 40). Loop over a copy, or build a new list with append.

Real output:
```
[40, 90]
```

After the fix it prints:
```
[90]
```

## Question 7 (Files, Output)

What is printed?

````python
for word in ['red', 'green', 'blue']:
    with open('colors.txt', 'w') as f:
        f.write(word + '\n')
with open('colors.txt') as f:
    print(f.read())
````

A) red, green, blue (each on its own line)

B) red only

C) blue only

D) redgreenblue

Answer: C

The file is opened with 'w' inside the loop, so every word erases the file first. Only the last word survives.

Real output:
```
blue
```

## Question 9 (Files, Output)

What is printed?

````python
with open('nums.txt', 'w') as f:
    f.write('5\n10\n')
with open('nums.txt') as f:
    a = f.readline().strip()
    b = f.readline().strip()
print(a + b, int(a) + int(b))
````

A) 15 15

B) 510 510

C) 510 15

D) A TypeError

Everything read from a file is a string, so '5' + '10' joins them into '510'. Only after int() do they add up to 15.

```
510 15
```

## Question 10 (Files, Debug)

diary.txt does not exist before the program runs. It should end with both days in it, but it crashes with FileExistsError. Which change fixes it?

````python
for day in ['Mon', 'Tue']:
    with open('diary.txt', 'x') as f:
        f.write(day + '\n')
print(open('diary.txt').read())
````

A) Change 'x' to 'w'

B) Change 'x' to 'a'

C) Change 'x' to 'r'

D) Remove the with and write f = open(...)

Answer: B

'x' only creates a new file, so the second day crashes. 'a' creates the file the first time and adds to the end after that. With 'w', only Tue would be left.

Real output:
```
FileExistsError: [Errno 17] File exists: 'diary.txt'
```

After the fix it prints:
```
Mon
Tue
```
