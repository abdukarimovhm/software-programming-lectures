## Question 1 (Strings, Output)

What is printed?

````python
s = 'Python'
print(s[0], s[-1])
print(s[1:4])
````

A) P n, then ytho

B) P y, then yth

C) P n, then yth

D) n P, then Pyt

Answer: C

s[-1] is the last character. In s[1:4] the end index 4 is not included, so you get indexes 1, 2 and 3.

Real output:
```
P n
yth
```

## Question 3 (Strings, Debug)

last_letter('INHA') should return 'A', but the program crashes with IndexError. Which change fixes it?

````python
def last_letter(word):
    return word[len(word)]

print(last_letter('INHA'))
````

A) return word[len(word) + 1]

B) return word[0]

C) return word[len(word) - 1]

D) return word(len(word))

Answer: C

The indexes of a 4-letter word go from 0 to 3, so word[4] is out of range. The last index is len(word) - 1 (or just word[-1]).

Real output:
```
IndexError: string index out of range
```

After the fix it prints:
```
A
```

## Question 5 (Lists, Debug)

sort_scores([70, 95, 60]) should return [60, 70, 95], but it prints None. Which change fixes it?

````python
def sort_scores(scores):
    scores = scores.sort()
    return scores

print(sort_scores([70, 95, 60]))
````

A) Change line 2 to scores = sorted(scores)

B) Change line 2 to scores.sort (without ())

C) Change line 3 to return None

D) Change line 2 to scores = scores.reverse()

Answer: A

.sort() sorts the list in place and returns None, so scores becomes None. sorted(scores) returns a new sorted list.

Real output:
```
None
```

After the fix it prints:
```
[60, 70, 95]
```

## Question 7 (Files, Output)

What is printed?

````python
with open('memo.txt', 'w') as f:
    f.write('one\n')
with open('memo.txt', 'w') as f:
    f.write('two\n')
with open('memo.txt') as f:
    print(f.read())
````

A) one, then two

B) one only

C) An error, because memo.txt already exists

D) two only

Answer: D

Mode 'w' erases the file before writing, so the second 'w' removes one. To add to the end, use 'a'.

Real output:
```
two
```

## Question 8 (Files, Output)

What is printed?

````python
with open('abc.txt', 'w') as f:
    f.write('HelloWorld')
with open('abc.txt') as f:
    print(f.read(5))
    print(f.read(5))
````

## Question 9 (Files, Output)

There is no file called no_such_file.txt. What is printed?

````python
try:
    f = open('no_such_file.txt')
    print('opened')
except:
    print('Something went wrong.')
print('done')
````

A) opened, then done

B) Something went wrong., then done

C) Only Something went wrong.

D) The program crashes with FileNotFoundError

Answer: B

The default mode 'r' raises FileNotFoundError on a missing file. except catches it, so opened is skipped, and the program keeps going to done.

Real output:
```
Something went wrong.
done
```

## Question 10 (Files, Debug)

The program should print each name on its own line, but it prints MinaJunAziz. Which change fixes it?

````python
names = ['Mina', 'Jun', 'Aziz']
with open('names.txt', 'w') as f:
    for name in names:
        f.write(name)
with open('names.txt') as f:
    print(f.read())
````

A) Change 'w' to 'a'

B) Change print(f.read()) to print(f.readline())

C) Change f.write(name) to f.write(name + '\n')

D) Change f.write(name) to f.write(names)

Answer: C

write() doesn't add a newline for you, so you have to write '\n' yourself after each name.

Real output:
```
MinaJunAziz
```

After the fix it prints:
```
Mina
Jun
Aziz
```
