## 1. Foundations — Code Blocks, Indentation, `return` vs `print`

**Remember:**

- A **block** starts after a line that ends with `:` (`def`, `if`, `elif`, `else`, `for`, `while`).
- Every line indented under it (4 spaces) is **inside** that block. The first line back at the old indentation is **outside**, so the block has ended.
- Indentation is part of the code, not decoration. Moving a line left or right changes **when** and **how many times** it runs. Messy indentation gives `IndentationError`.
- `print` **shows** a value on the screen. `return` **hands a value back** to whoever called the function.
- A function with no `return` gives back `None`. So `x = f()` stores `None`, and `print(f())` prints whatever `f` printed, then `None`.

## 2. Basics — Types and a Few Arithmetic Operators

**Remember:**

## 5. `for` Loops and `range`

**Remember:**

- `range(stop)` starts at **0** and stops **before** `stop`.
- `range(start, stop, step)`: with a negative step it counts down, and `stop` is still **not included**. If the range is empty, the loop body never runs.
- After the loop ends, the loop variable keeps its **last** value.
- Accumulators (`total = 0`, `count = 0`) are set up **before** the loop, not inside it.
- `for ch in "text"` visits each character, one at a time.

## 6. `while`, `break` / `continue`, Nested Loops

**Remember:**

- `while` keeps repeating **as long as** its condition is True, so something inside must eventually make it False.
- `break` exits the **innermost** loop immediately. `continue` skips the rest of the body and jumps to the **next** pass.
- In a `while` loop, if `continue` skips the line that updates the counter, you get an **infinite loop**.
- In nested loops, the inner loop runs **all the way through** for every pass of the outer loop.
- `n % 10` gives the last digit of `n`; `n //= 10` removes it.

## 7. Bitwise Operators and Number Systems

**Remember:**

- `0b` means binary, `0o` octal and `0x` hex. `bin()`, `oct()` and `hex()` convert a number to those forms, and `int("1011", 2)` converts back.
- `&` gives 1 where **both** bits are 1. `|` gives 1 where **either** is 1. `^` gives 1 where **exactly one** is 1.
- `~x` is `-x - 1`. `x << k` is `x * 2**k`. `x >> k` is `x // 2**k`.
- Permission flags: `p |= F` **gives** a flag, `p & F` **checks** it, `p ^= F` **flips** it and `p &= ~F` **removes** it.
- `and` / `or` work on **True/False**; `&` / `|` work on **bits**. And `^` is **not** power: `2 ^ 3` is `1`. Power is `**`.

### Q36 — Debug 🐞

**Goal:** `is_prime(n)` should return `True` only if `n` has no divisors other than 1 and itself.

a) What does it print **as written**?  b) Find the bug and fix it in the cell below, then re-run.

````python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
        else:
            return True

print(is_prime(9))   # expected False
print(is_prime(7))   # expected True
````
