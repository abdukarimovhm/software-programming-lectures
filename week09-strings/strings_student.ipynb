### 1.3 Slicing — a piece with `[start:end:step]`

| slice | meaning |
|---|---|
| `s[a:b]` | from index `a` **up to but not including** `b` |
| `s[:b]` | from the beginning |
| `s[a:]` | to the end |
| `s[:]` | a copy of the whole string |
| `s[a:b:2]` | every 2nd character |
| `s[::-1]` | the whole string **backwards** |

If the slice can't move from `start` toward `end` in the direction of `step`, you get the **empty string** `''` — no error!

### 1.6 String methods

Format: `string.method(...)`. A method **returns a new string** — the original never changes.

| method | returns | example |
|---|---|---|
| `s.upper()` / `s.lower()` | upper / lower case copy | `'Hi'.upper()` ➞ `'HI'` |
| `s.index(x)` | position of the first `x` — **error** if not found | `'abracadabra'.index('rac')` ➞ `2` |
| `s.count(x)` | how many times `x` appears | `'abracadabra'.count('a')` ➞ `5` |
| `s.strip()` | copy without spaces at both ends | `'  a b '.strip()` ➞ `'a b'` |
| `s.replace(a, b)` | copy with every `a` replaced by `b` | `'Hello'.replace('l', 'L')` ➞ `'HeLLo'` |
| `s.split(sep)` | a **list** of the pieces between `sep` | `'a,b'.split(',')` ➞ `['a', 'b']` |
| `s.split()` | split on spaces | `'hi  there'.split()` ➞ `['hi', 'there']` |

### Debug 2 — `capitalize_first`

**Expected:** `capitalize_first('python')` should return `'Python'` and `capitalize_first('inha')` should return `'Inha'`.

`TypeError: 'str' object does not support item assignment` — what does that tell you about strings? *(Don't use `.capitalize()` — build the new string yourself.)*

````python
def capitalize_first(s):
    s[0] = s[0].upper()
    return s

print(capitalize_first('python'))
print(capitalize_first('inha'))
````

### Debug 3 — `shout`

**Expected:** `shout('hello')` should return `'HELLO!'` and `shout('Inha')` should return `'INHA!'`.

No error — but the letters stay small. Look back at Predict 9.

````python
def shout(s):
    s.upper()
    return s + '!'

print(shout('hello'))
print(shout('Inha'))
````

### Debug 4 — `finds`

This is the search function from the slides, but someone "improved" it.

**Expected:** `finds('banana', 'n')` ➞ `2`, `finds('banana', 'b')` ➞ `0`, `finds('banana', 'z')` ➞ `-1`.

It returns `-1` for `'n'`. Trace it by hand: how many passes does the loop actually make?

````python
def finds(word, letter):
    index = 0
    while index < len(word):
        if word[index] == letter:
            return index
        else:
            return -1
        index += 1

print(finds('banana', 'n'))
print(finds('banana', 'b'))
print(finds('banana', 'z'))
````
