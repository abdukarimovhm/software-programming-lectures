## Question 5 (Compare two loops)

Loop A and Loop B are shown below. Which statement is true?

````python
for i in range(1, 6):
    print(i, end=" ")
````

````python
i = 1
while i < 6:
    print(i, end=" ")
    i += 1
````

- **A)** Loop A prints `1 2 3 4 5` and Loop B prints `1 2 3 4 5 6`
- **B)** Loop B never stops (infinite loop)
- **C)** They print exactly the same thing: `1 2 3 4 5`
- **D)** Loop A prints `1 2 3 4 5 6` and Loop B prints `1 2 3 4 5`

## Question 10 (What does it do?)

In plain English, what does `mystery(n)` calculate, and what is printed?

````python
def mystery(n):
    r = 0
    for i in range(1, n + 1):
        r += i * i
    return r

print(mystery(3))
````

- **A)** The square of the sum from 1 to n; prints `36`
- **B)** n squared; prints `9`
- **C)** The sum of the squares from 1 to n; prints `14`
- **D)** The sum from 1 to n; prints `6`

## Question 11 (Count the lines)

- **A)** 6

- **C)** 16
- **D)** 4

## Question 12 (Full output)

What is the full output (every line, in order)?

````python
def show(x):
    print("show", x)
    return x * 2

total = 0
for i in range(3):
    total += show(i)
print(total)
````

- **A)** `6` only
- **B)** `6`, then `show 0`, `show 1`, `show 2` (each on its own line)
- **C)** `show 0`, `show 1`, `show 2`, `3` (each on its own line)
- **D)** `show 0`, `show 1`, `show 2`, `6` (each on its own line)
