## Question 13

```python
def first_multiple(k, limit):
    for n in range(1, limit):
        if n % k == 0:
            return n
        return -1

print(first_multiple(1, 10))
print(first_multiple(3, 10))
```

- **A)** 1, then -1
- **B)** 1, then 3
- **C)** 3, then 3
- **D)** -1, then -1

## Question 15

```python
def digit_count(n):
    count = 0
    while n > 0:
        n //= 10
        count += 1
    return count

total = 0
for x in range(0, 12, 5):
    total += digit_count(x)
print(total)
```

- **A)** 4
- **B)** 3
- **C)** 6
- **D)** 15

# Part 2: Debugging

Each function below has **exactly one bug**. You can run the code to see the wrong result. Pick the option that describes the bug **and** the correct fix. Watch out for options that fix only one input or that sound right but change the wrong thing!

## Question 16 🐞

Expected `12` (2 + 4 + 6) but the program prints `6`. What is the bug and the fix?

```python
def total_evens(n):
    total = 0
    for i in range(2, n):
        if i % 2 == 0:
            total += i
    return total

print(total_evens(6))   # expected 12 (2 + 4 + 6), but prints 6
```

- **A)** Change `i % 2 == 0` to `i % 2 == 1`
- **B)** Move `return total` inside the loop
- **C)** Change `range(2, n)` to `range(2, n + 1)  ` so that `n` is included
- **D)** Change `total = 0` to `total = 2`

## Question 17 🐞

Expected `24` (1 × 2 × 3 × 4) but the program prints `0`. What is the bug and the fix?

```python
def product_to_n(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result

print(product_to_n(4))   # expected 24, but prints 0
```

- **A)** Change `range(1, n + 1)` to `range(n)`
- **B)** Change `result = 0` to `result = 1`
- **C)** Change `result * i` to `result + i`
- **D)** Indent `return result` inside the loop

## Question 18 🐞

Expected `6` (the first square strictly **greater** than 25) but the program prints `5`. What is the bug and the fix?

```python
def first_square_over(limit):
    for n in range(1, limit + 1):
        if n * n >= limit:
            return n

print(first_square_over(25))   # expected 6 (6 * 6 = 36 is the first square GREATER than 25), but prints 5
```

- **A)** Change `>=` to `>`
- **B)** Change `range(1, limit + 1)` to `range(limit)`
- **C)** Change `return n` to `return n + 1`  (this fixes only this one input, not the logic)
- **D)** Unindent `return n` so it runs after the loop

## Question 19 🐞

Expected `10` (9 + 1) but the program crashes with `TypeError`. What is the bug and the fix?

```python
def sum_odds(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 == 1:
            total += i
    print(total)

answer = sum_odds(5)
print(answer + 1)   # expected 10, but crashes
```

- **A)** The loop never runs; change `range(1, n + 1)` to `range(n)`
- **B)** `i % 2 == 1` is wrong; it should be `== 0`
- **C)** `answer + 1` is wrong; use `answer += 1`
- **D)** The function prints but never returns; replace `print(total)` with `return total`
