### 1.1 Opening a file — `open(filename, mode)`

`f = open('output.txt', 'w')` gives you a **file object** `f`. You use `f` to write or read, and you call `f.close()` when you're done. The **mode** says what you want to do:

| mode | meaning | if the file is **missing** | if the file **exists** |
|---|---|---|---|
| `'r'` *(default)* | read | 💥 `FileNotFoundError` | read it from the start |
| `'w'` | write | create it | ⚠️ **erase everything**, then write |
| `'a'` | append | create it | add to the **end** |
| `'x'` | create | create it | 💥 `FileExistsError` |

`open('output.txt')` with no mode is the same as `open('output.txt', 'r')`.

The demo is the example from the slides: `'w'`, then `'a'`, then `'w'` again. Before you run it, guess what is left at the end.

````python
f = open('output.txt', 'w')
f.write('This is the first line in the file\n')
f.close()
show('output.txt')

f = open('output.txt', 'a')
f.write('This is the 2nd line in the file\n')
f.close()
show('output.txt')

f = open('output.txt', 'w')
f.write('Is this the 3rd line in the file?\n')
f.close()
show('output.txt')
````

### 1.2 Writing — `f.write(string)`

Three rules to remember:

1. `write()` adds **no** newline. If you want a new line, write `'\n'` yourself.
2. `write()` only takes **strings**. Turn numbers into strings first with `str(...)`.
3. `write()` **returns** how many characters it wrote (that's the `35` you saw in the lecture's shell).

````python
f = open('demo.txt', 'w')
n = f.write('score: ' + str(95) + '\n')
f.write('no newline here')
f.write('...so this sticks to it\n')
f.close()

print(n)
show('demo.txt')
````

### 1.3 Reading — four ways

| code | gives you |
|---|---|
| `f.read()` | the whole rest of the file, as **one string** |
| `f.read(n)` | the next `n` characters |
| `f.readline()` | the next line, **including** its `'\n'` |
| `for line in f:` | each remaining line, **including** its `'\n'` |

Everything you read is a **string**, even if it looks like a number: use `int(...)` to get a number back. `line.strip()` removes the `'\n'` at the end.

````python
f = open('harrypotter.txt')
print(f.read())
f.close()

f = open('harrypotter.txt')
first = f.readline()
f.close()
print(repr(first))
print(first.strip())
````

### 1.4 Safe files — `with` and `try`

- `with open(...) as f:` closes the file **for you** at the end of the block, even if an error happens. Prefer it to `open` + `close`.
- `try` / `except` catches errors, for example a file that doesn't exist, so the program doesn't crash.

This is the code from the slides.

````python
try:
    with open('harrypotter.txt') as f:
        for line in f:
            print(line, end='')
except:
    print('File reading error!')

print(f.closed)

try:
    fin = open('bad_file')
except:
    print('Something went wrong.')
````

### 1.5 Files and folders — the `os` module

| code | what it does |
|---|---|
| `os.getcwd()` | the folder your program is running in |
| `os.listdir(folder)` | a list of the **names** in a folder (files and folders) |
| `os.path.join(folder, name)` | glue them into one path, like `'notes/todo.txt'` |
| `os.path.exists(path)` | `True` if the file **or** folder is there |
| `os.path.isfile(path)` / `os.path.isdir(path)` | `True` only for a file / only for a folder |
| `os.path.abspath(name)` | the full path, from the top of the disk |
| `os.remove(path)` / `os.rmdir(folder)` | delete a file / delete an **empty** folder |
| `os.mkdir(folder)` | make a new folder (the opposite of `rmdir`) |

````python
import os
print(os.getcwd())                       # in Colab: /content

if not os.path.exists('notes'):
    os.mkdir('notes')
path = os.path.join('notes', 'todo.txt')
with open(path, 'w') as f:
    f.write('finish the proposal\n')

print(path, os.path.abspath(path))
print(os.listdir('notes'))
print(os.path.isfile(path), os.path.isdir('notes'), os.path.isdir(path))

os.remove(path)                          # delete the file first...
os.rmdir('notes')                        # ...because rmdir needs an EMPTY folder
print(os.path.exists('notes'))
````

## 6. Debrief — The Three File Traps

| trap | what happens | do this instead |
|---|---|---|
| `open(name, 'w')` | the file is **erased** first, even inside a loop | open it once before the loop, or use `'a'` to add |
| `f.write(x)` | adds **no** `'\n'`, and crashes if `x` is a number | `f.write(str(x) + '\n')` |
| reading numbers back | every line is a **string** like `'70\n'` | `int(line.strip())` |

Use `with open(...) as f:` so the file always closes, and `try` / `except` when the file might be missing.

- `with` closes the file for you, and `try` / `except` handles `FileNotFoundError` and `FileExistsError`
- `os.getcwd`, `os.listdir`, `os.path.join`, `os.path.exists`, `os.path.isfile`, `os.path.isdir`, `os.remove`, `os.mkdir` and `os.rmdir`
