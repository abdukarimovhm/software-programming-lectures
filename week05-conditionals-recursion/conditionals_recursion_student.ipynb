### Concept Recap (quick — no slides needed today)

- **Compound statement** — a header ending in `:` plus an indented body (`def`, `if`, `else`, ...).
- **Indentation is scope** — a line that is *not* indented under the `if` always runs.
- **`elif` order matters** — conditions are checked top to bottom and the **first** true one wins; the rest are skipped, even if they would also be true.
- **A variable assigned only inside one branch** may not exist if that branch is skipped → `NameError` / `UnboundLocalError`.
- **A function with an uncovered path** doesn't crash — it silently returns `None`.
- **`0 < x < 10`** is Python's concise form of `0 < x and x < 10` — prefer it over nesting.
- **`pass`** — a do-nothing placeholder body.

## 5. Recursion — A Function That Calls Itself

A recursive function solves a problem by reducing it to a **smaller version of the same problem**. Every recursive function needs two parts:

- **Base case** — a case so simple it is answered directly, with no further call.
- **Recursive case** — calls the function again on something *closer to the base case*.

```python
def i_am_recursive(x):
    if x is the simplest case:          # base case
        return the answer directly
    return ... i_am_recursive(smaller x) ...   # recursive case
```

If the recursive call is **not** closer to the base case (or there is no base case), the function never stops — Python eventually gives up with a `RecursionError`.

### Tracing `factorial(3)` by hand

`factorial(n)` is `n * (n-1) * ... * 1`, with `factorial(0)` = `1`. Here is how the calls stack up and then unwind:

```
factorial(3)
= 3 * factorial(2)
= 3 * (2 * factorial(1))
= 3 * (2 * (1 * factorial(0)))
= 3 * (2 * (1 * 1))        <- base case reached, now results flow back
= 3 * (2 * 1)
= 3 * 2
= 6
```

**Rule of thumb:** to trust a recursive function, check (1) the base case is right, (2) each call really moves toward it, and (3) *if* the smaller call is correct, the combining step makes the bigger one correct.

This one *has* a base case, yet it still never finishes. Trace `factorial(2)` by hand: what does it call?

It crashes with a `TypeError` that mentions `NoneType`. Where did a `None` come from — what does `power` hand back when `exp` is not `0`?

### Step 0 — Play the starter

Run the cell below, type `rock`, `paper` or `scissors`, and see what happens. **Run it several times** — the computer's move is random.

````python
import random

def play():
    player = input("rock, paper or scissors? ")

    number = random.randint(1, 3)
    if number == 1:
        computer = "rock"
    elif number == 2:
        computer = "paper"
    else:
        computer = "scissors"
    print("Computer chose:", computer)

    if player == computer:
        print("Draw!")
    elif (player == "rock" and computer == "scissors") or (player == "scissors" and computer == "paper") or (player == "paper" and computer == "rock"):
        print("You win!")
    else:
        print("You lose!")

play()
````
