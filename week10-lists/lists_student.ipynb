### Cheat sheet

| want | code | changes the list? | returns |
|---|---|---|---|
| add to the end | `x.append(item)` | ✅ | `None` |
| add at a position | `x.insert(i, item)` | ✅ | `None` |
| remove by value (first match) | `x.remove(item)` | ✅ | `None` (ValueError if missing) |
| remove by position | `x.pop()` / `x.pop(i)` | ✅ | **the removed item** |
| sort / reverse in place | `x.sort()` / `x.reverse()` | ✅ | `None` |
| a sorted **copy** | `sorted(x)` | ❌ | new list |
| how many / where | `x.count(item)` / `x.index(item)` | ❌ | int |
| a real copy | `x.copy()` | ❌ | new list |
| size, biggest, smallest, total | `len(x)`, `max(x)`, `min(x)`, `sum(x)` | ❌ | a value |

### Debug 2 — `remove_negatives`

**Expected:** `remove_negatives([3, -1, 4])` ➞ `[3, 4]` and `remove_negatives([-1, -2, 3, -4])` ➞ `[3]`

The first one works, the second one doesn't. Trace the loop by hand: after `remove`, which item does the loop look at next?

````python
def remove_negatives(nums):
    for x in nums:
        if x < 0:
            nums.remove(x)
    return nums

print(remove_negatives([3, -1, 4]))
print(remove_negatives([-1, -2, 3, -4]))
````

### Debug 3 — `find_max`

**Expected:** `find_max([3, 9, 2])` ➞ `9` and `find_max([-5, -2, -9])` ➞ `-2`

No error, but the second answer is a number that isn't even in the list. *(Don't use `max()`, just fix the loop.)*

````python
def find_max(nums):
    biggest = 0
    for x in nums:
        if x > biggest:
            biggest = x
    return biggest

print(find_max([3, 9, 2]))
print(find_max([-5, -2, -9]))
````

## 5. Debrief — The Three List Traps

| trap | what happens | do this instead |
|---|---|---|
| `x = x.sort()` | list methods change the list and return **`None`** | `x.sort()` on its own line, or `y = sorted(x)` |
| `b = a` | one list, **two names**: changing `b` changes `a` | `b = a.copy()` |
| `remove` inside `for x in a` | the loop **skips** items | build a new list with `append` |

## What We Practiced

- Indexing from 0, negative indexes from the end, and the `IndexError` you get past the last item
- Slices `[start:end:step]` return a **list**, never crash, and can be empty
- `+` joins lists, `*` repeats them, and `in` / `not in` check for whole items
- Lists are **mutable**: `x[i] = …`, `x[a:b] = […]` and `del x[i]` change the list in place
- `sorted(x)` gives a new list, while `x.sort()` changes `x` and returns `None`
- `append`, `insert`, `remove`, `pop`, `count`, `index`, `reverse`, `clear` and `copy`
- `b = a` is a second name for the same list, and `a.copy()` is a real copy
